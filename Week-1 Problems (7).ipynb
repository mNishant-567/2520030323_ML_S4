```python
# Declare variables of type int, float, string, and boolean, and display their values along with their data types.
print("Program 1: Variables and Data Types\n")

my_integer = 10
my_float = 20.5
my_string = "Hello, Python!"
my_boolean = True

print("--- Initial Variable Declarations ---")
print(f"Value: {my_integer}, Type: {type(my_integer)}")
print(f"Value: {my_float}, Type: {type(my_float)}")
print(f"Value: {my_string}, Type: {type(my_string)}")
print(f"Value: {my_boolean}, Type: {type(my_boolean)}")
```

```python
# Modify the values of the variables and observe how their data types change.
# In Python, types are dynamic, so assigning a new value of a different type changes the variable's type.
print("\n--- Modifying Variable Values and Observing Type Changes ---")
my_integer = "Now I am a string"
my_float = False
my_string = 123.45
my_boolean = 99

print(f"Value: {my_integer}, Type: {type(my_integer)}")
print(f"Value: {my_float}, Type: {type(my_float)}")
print(f"Value: {my_string}, Type: {type(my_string)}")
print(f"Value: {my_boolean}, Type: {type(my_boolean)}")
```

```python
# Create your own variables representing a student's details and print their values and data types.
print("\n--- Student Details ---")
student_name = "Alice Smith"
student_age = 20
student_gpa = 3.85
is_enrolled = True
student_id = "S12345"
student_major = "Computer Science"

print(f"Student Name: {student_name}, Type: {type(student_name)}")
print(f"Student Age: {student_age}, Type: {type(student_age)}")
print(f"Student GPA: {student_gpa}, Type: {type(student_gpa)}")
print(f"Is Enrolled: {is_enrolled}, Type: {type(is_enrolled)}")
print(f"Student ID: {student_id}, Type: {type(student_id)}")
print(f"Student Major: {student_major}, Type: {type(student_major)}")
```

```python
# 1. Arithmetic Operators
a = 15
b = 4

print("--- Arithmetic Operators ---")
print(f"{a} + {b} = {a + b}")  # Addition
print(f"{a} - {b} = {a - b}")  # Subtraction
print(f"{a} * {b} = {a * b}")  # Multiplication
print(f"{a} / {b} = {a / b}")  # Division
print(f"{a} // {b} = {a // b}") # Floor Division
print(f"{a} % {b} = {a % b}")  # Modulus
print(f"{a} ** {b} = {a ** b}") # Exponentiation
```

```python
# 2. Comparison Operators
x = 10
y = 20

print("\n--- Comparison Operators ---")
print(f"{x} == {y}: {x == y}")
print(f"{x} != {y}: {x != y}")
print(f"{x} > {y}: {x > y}")
print(f"{x} < {y}: {x < y}")
print(f"{x} >= {y}: {x >= y}")
print(f"{x} <= {y}: {x <= y}")
```

```python
# 3. Logical Operators
t = True
f = False

print("\n--- Logical Operators ---")
print(f"{t} and {f}: {t and f}")
print(f"{t} or {f}: {t or f}")
print(f"not {t}: {not t}")
```

```python
# 4. Placement Eligibility Check
cgpa = 7.5
attendance = 80

# Eligibility criteria: CGPA >= 7.0 and Attendance >= 75%
is_eligible = (cgpa >= 7.0) and (attendance >= 75)

print("\n--- Placement Eligibility ---")
print(f"Student CGPA: {cgpa}")
print(f"Student Attendance: {attendance}%")
print(f"Is the student eligible for placement? {is_eligible}")
```
