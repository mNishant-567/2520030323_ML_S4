```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os

# -------------------------------------------------------------
# 2. Load the Dataset
# -------------------------------------------------------------
# Ensure you replace the filename with your exact file path if different
file_path = '/content/placement_predict_50k Dataset (1).csv'

try:
    df = pd.read_csv(file_path)
except FileNotFoundError:
    print(f"Error: File '{file_path}' not found. Please check the filename or upload it to Colab.")
    exit()

print("Dataset loaded successfully!")
display(df.head())

# -------------------------------------------------------------
# 3. Identify Features (Numerical vs Categorical)
# -------------------------------------------------------------
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns
categorical_features = df.select_dtypes(include=['object']).columns

print("Numerical Features:", list(numerical_features))
print("Categorical Features:", list(categorical_features))

# Select only numerical columns for correlation analysis
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()

# -------------------------------------------------------------
# 4. Correlation Heatmap
# -------------------------------------------------------------
corr_matrix = df[numerical_cols].corr()
display(corr_matrix)

output_dir = "/content/"
heatmap_path = os.path.join(output_dir, "correlation_heatmap.png")

plt.figure(figsize=(12, 10)) # Increased figure size for better visibility
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.savefig(heatmap_path, dpi=300)
plt.close()
print(f"Exported heatmap to: {heatmap_path}")

# -------------------------------------------------------------
# 5. Boxplots: Numerical Features vs PlacementStatus
# -------------------------------------------------------------
target_col = "PlacementStatus"

if target_col in df.columns:
    for col in numerical_cols:
        plt.figure(figsize=(6, 5))

        # Using x as categorical and y as numerical
        sns.boxplot(
            x=target_col,
            y=col,
            data=df,
            palette="Set2",
            hue=target_col,
            legend=False
        )
        plt.title(f'Distribution of {col} by {target_col}')
        plt.xlabel(target_col)
        plt.ylabel(col)
        plt.grid(True)
        plt.show()
```
